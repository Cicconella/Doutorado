```python

import numpy as np
from PIL import Image
from skimage.measure import shannon_entropy
from skimage.feature import graycomatrix, graycoprops
from skimage.util import img_as_ubyte
from skimage.filters import sobel
from scipy.stats import kurtosis, skew

def remover_bordas_pretas(imagem_np, limiar_preto=5):
    altura, largura = imagem_np.shape

    linhas_topo = 0
    for i in range(altura):
        if np.all(imagem_np[i, :] <= limiar_preto):
            linhas_topo += 1
        else:
            break

    linhas_base = 0
    for i in range(altura - 1, -1, -1):
        if np.all(imagem_np[i, :] <= limiar_preto):
            linhas_base += 1
        else:
            break

    colunas_esq = 0
    for j in range(largura):
        if np.all(imagem_np[:, j] <= limiar_preto):
            colunas_esq += 1
        else:
            break

    colunas_dir = 0
    for j in range(largura - 1, -1, -1):
        if np.all(imagem_np[:, j] <= limiar_preto):
            colunas_dir += 1
        else:
            break

    imagem_recortada = imagem_np[
        linhas_topo:altura - linhas_base if linhas_base > 0 else altura,
        colunas_esq:largura - colunas_dir if colunas_dir > 0 else largura
    ]

    return imagem_recortada, {
        "linhas_pretas_topo": linhas_topo,
        "linhas_pretas_base": linhas_base
    }, {
        "colunas_pretas_esquerda": colunas_esq,
        "colunas_pretas_direita": colunas_dir
    }

def extrair_medidas_estatisticas(imagem_np):
    return {
        "media": np.mean(imagem_np),
        "mediana": np.median(imagem_np),
        "desvio_padrao": np.std(imagem_np),
        "minimo": np.min(imagem_np),
        "maximo": np.max(imagem_np),
        "entropia": shannon_entropy(imagem_np),
        "assimetria": skew(imagem_np.flatten()),
        "curtose": kurtosis(imagem_np.flatten())
    }

def extrair_textura_glcm(imagem_np):
    imagem_ubyte = img_as_ubyte(imagem_np)
    glcm = graycomatrix(imagem_ubyte, distances=[1], angles=[0], levels=256, symmetric=True, normed=True)
    return {
        "contraste": graycoprops(glcm, 'contrast')[0, 0],
        "homogeneidade": graycoprops(glcm, 'homogeneity')[0, 0],
        "energia": graycoprops(glcm, 'energy')[0, 0],
        "correlacao": graycoprops(glcm, 'correlation')[0, 0]
    }

def extrair_bordas(imagem_np):
    bordas = sobel(imagem_np)
    return {
        "media_bordas": np.mean(bordas),
        "desvio_bordas": np.std(bordas)
    }

def pipeline_extracao_features(caminho_imagem):
    imagem = Image.open(caminho_imagem).convert("L")
    imagem_np = np.array(imagem)
    imagem_np_recortada, linhas_pretas, colunas_pretas = remover_bordas_pretas(imagem_np)

    medidas = {}
    medidas.update(extrair_medidas_estatisticas(imagem_np_recortada))
    medidas.update(extrair_textura_glcm(imagem_np_recortada))
    medidas.update(extrair_bordas(imagem_np_recortada))
    medidas.update(linhas_pretas)
    medidas.update(colunas_pretas)

    return medidas
```

# Calculando para várias clinicas

```python
import pandas as pd
import os
from PIL import Image

def calcular_estatisticas_clinica(caminho_lista_exames, end_pans, limite_exames=None):
    """
    Calcula as estatísticas de imagens de uma clínica.
    
    Parâmetros:
        caminho_lista_exames (str): Caminho para o arquivo CSV contendo os exames.
        end_pans (str): Caminho para a pasta contendo as imagens panorâmicas.
        limite_exames (int, opcional): Número máximo de exames a processar (para testes).
        
    Retorna:
        DataFrame: Estatísticas extraídas para cada exame.
    """
    exames = pd.read_csv(caminho_lista_exames)
    
    if limite_exames is not None:
        exames = exames.head(limite_exames)

    resultados = []

    for exame in exames.cd_exame:
        nome_arquivo = f"{exame.upper()}-pan.jpg"
        caminho_imagem = os.path.join(end_pans, nome_arquivo)

        try:
            imagem = Image.open(caminho_imagem)
            estatisticas = pipeline_extracao_features(caminho_imagem)
            estatisticas["cd_exame"] = exame
            resultados.append(estatisticas)
        except Exception as e:
            print(f"Erro ao processar {exame}: {e}")

    df_estatisticas = pd.DataFrame(resultados)

    # Reordenar colunas: cd_exame primeiro
    colunas = ['cd_exame'] + [col for col in df_estatisticas.columns if col != 'cd_exame']
    df_estatisticas = df_estatisticas[colunas]

    return df_estatisticas


def gerar_caminhos_clinicas(lista_clinicas, base_path, sufixo="_exames.csv"):
    """
    Gera os caminhos dos arquivos CSV das clínicas automaticamente.
    
    Parâmetros:
        lista_clinicas (list): Lista com os nomes das clínicas.
        base_path (str): Caminho base onde estão os arquivos.
        sufixo (str): Sufixo que será adicionado ao nome da clínica (default "_exames.csv").
    
    Retorna:
        dict: Dicionário {nome_clinica: caminho_csv}.
    """
    caminhos = {}
    for clinica in lista_clinicas:
        caminho = os.path.join(base_path, f"{clinica}_exames.csv")
        caminhos[clinica] = caminho
    return caminhos
```

```python
# Lista simples das clínicas
lista_clinicas = ["Alphaville", "Ibirapuera", "Ipiranga", "Jardins", "Jd._America"]

# Caminho base onde estão os CSVs
base_path = "/home/anaciconelle/Desktop/Doutorado/exames_por_origem"

# Gerar automaticamente os caminhos
clinicas = gerar_caminhos_clinicas(lista_clinicas, base_path)

# Dicionário para guardar os DataFrames de cada clínica
estatisticas_clinicas = {}

for clinica, caminho_csv in clinicas.items():
    print(f"Processando clínica: {clinica}")
    df_estatisticas = calcular_estatisticas_clinica(caminho_csv, end_pans, limite_exames=100)
    estatisticas_clinicas[clinica] = df_estatisticas
```

```python
def comparar_df_por_nomes(nome1, nome2, dicionario_df, plot_boxplots=True, plot_pca=True):
    """
    Compara dois grupos de estatísticas com base em seus nomes dentro de um dicionário.
    
    Parâmetros:
        nome1, nome2: Nomes dos grupos no dicionário (ex: 'Tucuruvi', 'Morumbi')
        dicionario_df: Dict com DataFrames das estatísticas para cada grupo
        plot_boxplots: Se True, plota boxplots para cada variável
        plot_pca: Se True, plota PCA
        
    Retorna:
        DataFrame com médias e p-valores dos testes estatísticos
    """
    df1 = dicionario_df[nome1].copy()
    df2 = dicionario_df[nome2].copy()
    
    df1['grupo'] = nome1
    df2['grupo'] = nome2

    df_total = pd.concat([df1, df2], ignore_index=True)

    colunas_excluir = ['cd_exame', 'grupo']
    variaveis_numericas = df_total.drop(columns=colunas_excluir, errors='ignore')

    resultados = []

    for coluna in variaveis_numericas.columns:
        grupo1 = df1[coluna]
        grupo2 = df2[coluna]
        
        media1 = np.mean(grupo1)
        media2 = np.mean(grupo2)

        # Teste t
        t_stat, p_ttest = ttest_ind(grupo1, grupo2, equal_var=False)
        
        # Teste Mann-Whitney
        u_stat, p_mw = mannwhitneyu(grupo1, grupo2, alternative='two-sided')
        
        resultados.append({
            'variavel': coluna,
            f'media_{nome1}': media1,
            f'media_{nome2}': media2,
            'p_ttest': p_ttest,
            'p_mannwhitney': p_mw
        })
    
    df_resultados = pd.DataFrame(resultados)

    # --- Boxplots (opcional)
    if plot_boxplots:
        for var in variaveis_numericas.columns:
            plt.figure(figsize=(6,4))
            sns.boxplot(x='grupo', y=var, data=df_total)
            plt.title(f'Comparação de {var}')
            plt.grid(True)
            plt.show()
    
    # --- PCA (opcional)
    if plot_pca:
        scaler = StandardScaler()
        variaveis_normalizadas = scaler.fit_transform(variaveis_numericas)
        
        pca = PCA(n_components=2)
        pca_resultado = pca.fit_transform(variaveis_normalizadas)
        
        df_pca = pd.DataFrame(pca_resultado, columns=['PC1', 'PC2'])
        df_pca['grupo'] = df_total['grupo'].values
        
        plt.figure(figsize=(8,6))
        sns.scatterplot(data=df_pca, x='PC1', y='PC2', hue='grupo', s=100)
        plt.title(f'PCA - {nome1} vs {nome2}')
        plt.grid(True)
        plt.show()

        print("Variância explicada pelo PCA:", pca.explained_variance_ratio_)

    return df_resultados

```

```python
# Função para comparação múltipla entre clínicas com heatmap e PCA scatterplots

def comparar_multiplos_grupos(lista_clinicas, estatisticas_clinicas, salvar_path="/mnt/data", salvar_csv=True, salvar_fig=True):
    import itertools
    import os

    # Inicializa matrizes de p-valores
    matriz_ttest = pd.DataFrame(index=lista_clinicas, columns=lista_clinicas, dtype=float)
    matriz_mw = pd.DataFrame(index=lista_clinicas, columns=lista_clinicas, dtype=float)
    pca_todos = []

    for clinica1, clinica2 in itertools.combinations(lista_clinicas, 2):
        try:
            resultado = comparar_df_por_nomes(clinica1, clinica2, estatisticas_clinicas, plot_pca=False, plot_boxplots=False)

            # Média dos p-valores
            p_ttest_medio = resultado['p_ttest'].mean()
            p_mw_medio = resultado['p_mannwhitney'].mean()

            matriz_ttest.loc[clinica1, clinica2] = p_ttest_medio
            matriz_ttest.loc[clinica2, clinica1] = p_ttest_medio
            matriz_mw.loc[clinica1, clinica2] = p_mw_medio
            matriz_mw.loc[clinica2, clinica1] = p_mw_medio

            # PCA
            df1 = estatisticas_clinicas[clinica1].copy()
            df2 = estatisticas_clinicas[clinica2].copy()
            df1['grupo'] = clinica1
            df2['grupo'] = clinica2
            df_total = pd.concat([df1, df2], ignore_index=True)

            colunas_excluir = ['cd_exame', 'grupo']
            variaveis_numericas = df_total.drop(columns=colunas_excluir, errors='ignore')

            scaler = StandardScaler()
            variaveis_normalizadas = scaler.fit_transform(variaveis_numericas)

            pca = PCA(n_components=2)
            pca_resultado = pca.fit_transform(variaveis_normalizadas)

            df_pca = pd.DataFrame(pca_resultado, columns=['PC1', 'PC2'])
            df_pca['grupo'] = df_total['grupo'].values
            df_pca['par'] = f"{clinica1} vs {clinica2}"

            pca_todos.append(df_pca)

        except Exception as e:
            print(f"Erro ao comparar {clinica1} e {clinica2}: {e}")

    # Salvar CSVs
    if salvar_csv:
        ttest_path = os.path.join(salvar_path, "heatmap_ttest.csv")
        mw_path = os.path.join(salvar_path, "heatmap_mannwhitney.csv")
        matriz_ttest.to_csv(ttest_path)
        matriz_mw.to_csv(mw_path)

    # Heatmaps
    plt.figure(figsize=(10, 8))
    sns.heatmap(matriz_ttest.astype(float), annot=True, cmap='coolwarm', square=True)
    plt.title('Heatmap dos p-valores médios (t-test)')
    if salvar_fig:
        plt.savefig(os.path.join(salvar_path, "heatmap_ttest.png"))
    plt.show()

    plt.figure(figsize=(10, 8))
    sns.heatmap(matriz_mw.astype(float), annot=True, cmap='coolwarm', square=True)
    plt.title('Heatmap dos p-valores médios (Mann-Whitney)')
    if salvar_fig:
        plt.savefig(os.path.join(salvar_path, "heatmap_mannwhitney.png"))
    plt.show()

    # Scatterplots PCA
    if pca_todos:
        df_pca_todos = pd.concat(pca_todos, ignore_index=True)
        g = sns.FacetGrid(df_pca_todos, col="par", hue="grupo", col_wrap=3, height=4, sharex=False, sharey=False)
        g.map_dataframe(sns.scatterplot, x="PC1", y="PC2")
        g.add_legend()
        plt.subplots_adjust(top=0.9)
        g.fig.suptitle('PCA entre pares de clínicas')
        if salvar_fig:
            g.savefig(os.path.join(salvar_path, "pca_pairs.png"))
        plt.show()

    return matriz_ttest, matriz_mw
```
